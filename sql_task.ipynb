### Subtask:
Perform at least 10 SQL queries on the 'Authors' and 'Books' tables, demonstrating `JOIN`, `PRIMARY KEY` (implicitly), `WHERE`, `AND`, `GROUP BY`, and `HAVING` clauses.


**Reasoning**:
The subtask requires executing multiple SQL queries on the 'Authors' and 'Books' tables, demonstrating various SQL clauses. This step will re-create the database, insert sample data, and then execute each query as specified, printing the query and its results.



```
import sqlite3
import random
import datetime

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Re-create Authors table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Authors (
    author_id INTEGER PRIMARY KEY,
    first_name TEXT NOT NULL,
    last_name TEXT NOT NULL,
    DOB TEXT
);
''')

# Re-create Books table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Books (
    book_id INTEGER PRIMARY KEY,
    book_name TEXT NOT NULL,
    author_id INTEGER,
    FOREIGN KEY (author_id) REFERENCES Authors(author_id)
);
''')

# Generate sample data for Authors (50 records)
author_data = []
first_names = ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan', 'Fiona', 'George', 'Hannah', 'Ian', 'Julia']
last_names = ['Smith', 'Jones', 'Williams', 'Brown', 'Davis', 'Miller', 'Wilson', 'Moore', 'Taylor', 'Anderson']

for i in range(50):
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)
    year = random.randint(1900, 2000)
    month = random.randint(1, 12)
    day = random.randint(1, 28) # To avoid issues with months having fewer days
    dob = f"{year:04d}-{month:02d}-{day:02d}"
    author_data.append((first_name, last_name, dob))

cursor.executemany("INSERT INTO Authors (first_name, last_name, DOB) VALUES (?, ?, ?)", author_data)

# Generate sample data for Books (at least 50 records)
book_data = []
book_titles = [
    'The Great Adventure',
    'Mystery of the Old House',
    'Stars Beyond Reach',
    'Whispers in the Wind',
    'Journey to the Unknown',
    'Shadows of the Past',
    'The Last Chapter',
    'Echoes of Tomorrow',
    'A Silent Promise',
    'Beneath the Surface'
]

min_author_id = 1
max_author_id = 50

for i in range(50):
    book_name = f"{random.choice(book_titles)} {i+1}"
    author_id = random.randint(min_author_id, max_author_id)
    book_data.append((book_name, author_id))

cursor.executemany("INSERT INTO Books (book_name, author_id) VALUES (?, ?)", book_data)

# Commit the changes
conn.commit()

print("Database setup complete: Tables re-created and 50 sample records inserted into 'Authors' and 'Books'.\n")

# --- SQL Queries ---

# Query 1: Demonstrate JOIN - Select all books along with the first and last name of their authors.
query_1 = """SELECT b.book_name, a.first_name, a.last_name FROM Books b JOIN Authors a ON b.author_id = a.author_id;"""
cursor.execute(query_1)
results_1 = cursor.fetchall()
print("Query 1: Demonstrate JOIN (Books with Author Names)\nSQL:\n" + query_1)
print("Results (first 5):", results_1[:5])
print("\n")

# Query 2: Demonstrate WHERE - Select all authors born after a specific year (e.g., 1950).
specific_year = 1950
query_2 = f"""SELECT first_name, last_name, DOB FROM Authors WHERE SUBSTR(DOB, 1, 4) > '{specific_year}';"""
cursor.execute(query_2)
results_2 = cursor.fetchall()
print(f"Query 2: Demonstrate WHERE (Authors born after {specific_year})\nSQL:\n" + query_2)
print("Results (first 5):", results_2[:5])
print("\n")

# Query 3: Demonstrate WHERE and AND - Select books by a specific author ('Alice' 'Smith') and born before a specific year.
specific_first_name = 'Alice'
specific_last_name = 'Smith'
birth_year_before = 1980
query_3 = f"""SELECT b.book_name, a.first_name, a.last_name, a.DOB FROM Books b JOIN Authors a ON b.author_id = a.author_id WHERE a.first_name = '{specific_first_name}' AND a.last_name = '{specific_last_name}' AND SUBSTR(a.DOB, 1, 4) < '{birth_year_before}';"""
cursor.execute(query_3)
results_3 = cursor.fetchall()
print(f"Query 3: Demonstrate WHERE and AND (Books by {specific_first_name} {specific_last_name} born before {birth_year_before})\nSQL:\n" + query_3)
print("Results (first 5):", results_3[:5])
print("\n")

# Query 4: Demonstrate GROUP BY - Count the number of books written by each author, showing the author's full name and the count.
query_4 = """SELECT a.first_name, a.last_name, COUNT(b.book_id) AS book_count FROM Authors a LEFT JOIN Books b ON a.author_id = b.author_id GROUP BY a.author_id, a.first_name, a.last_name ORDER BY book_count DESC;"""
cursor.execute(query_4)
results_4 = cursor.fetchall()
print("Query 4: Demonstrate GROUP BY (Book count per author)\nSQL:\n" + query_4)
print("Results (first 5):", results_4[:5])
print("\n")

# Query 5: Demonstrate GROUP BY and HAVING - Find authors who have written more than a certain number of books (e.g., 5 books).
min_books = 5
query_5 = f"""SELECT a.first_name, a.last_name, COUNT(b.book_id) AS book_count FROM Authors a JOIN Books b ON a.author_id = b.author_id GROUP BY a.author_id, a.first_name, a.last_name HAVING book_count > {min_books} ORDER BY book_count DESC;"""
cursor.execute(query_5)
results_5 = cursor.fetchall()
print(f"Query 5: Demonstrate GROUP BY and HAVING (Authors with more than {min_books} books)\nSQL:\n" + query_5)
print("Results:", results_5)
print("\n")

# Query 6: Demonstrate PRIMARY KEY (implicit) - Select a specific author by their author_id.
specific_author_id = 15 # Choose an existing ID
query_6 = f"""SELECT * FROM Authors WHERE author_id = {specific_author_id};"""
cursor.execute(query_6)
results_6 = cursor.fetchone()
print(f"Query 6: Demonstrate PRIMARY KEY (Select Author with ID {specific_author_id})\nSQL:\n" + query_6)
print("Results:", results_6)
print("\n")

# Query 7: Demonstrate ORDER BY - List all authors, ordered by their date of birth (DOB) in ascending order.
query_7 = """SELECT first_name, last_name, DOB FROM Authors ORDER BY DOB ASC;"""
cursor.execute(query_7)
results_7 = cursor.fetchall()
print("Query 7: Demonstrate ORDER BY (Authors by DOB)\nSQL:\n" + query_7)
print("Results (first 5):", results_7[:5])
print("\n")

# Query 8: Demonstrate COUNT(*) - Count the total number of authors in the Authors table.
query_8 = """SELECT COUNT(*) FROM Authors;"""
cursor.execute(query_8)
results_8 = cursor.fetchone()[0]
print("Query 8: Demonstrate COUNT(*) (Total Authors)\nSQL:\n" + query_8)
print("Results:", results_8)
print("\n")

# Query 9: Demonstrate DISTINCT - List all unique last names of authors.
query_9 = """SELECT DISTINCT last_name FROM Authors ORDER BY last_name;"""
cursor.execute(query_9)
results_9 = cursor.fetchall()
print("Query 9: Demonstrate DISTINCT (Unique Last Names)\nSQL:\n" + query_9)
print("Results:", results_9)
print("\n")

# Query 10: Demonstrate LIKE with WHERE - Find books whose names contain a specific keyword (e.g., 'Mystery').
keyword = 'Mystery'
query_10 = f"""SELECT book_name FROM Books WHERE book_name LIKE '%{keyword}%';"""
cursor.execute(query_10)
results_10 = cursor.fetchall()
print(f"Query 10: Demonstrate LIKE with WHERE (Books containing '{keyword}')\nSQL:\n" + query_10)
print("Results:", results_10)
print("\n")

# Query 11: Demonstrate BETWEEN with WHERE - Select authors born between two specific years.
year_start = 1960
year_end = 1980
query_11 = f"""SELECT first_name, last_name, DOB FROM Authors WHERE SUBSTR(DOB, 1, 4) BETWEEN '{year_start}' AND '{year_end}' ORDER BY DOB;"""
cursor.execute(query_11)
results_11 = cursor.fetchall()
print(f"Query 11: Demonstrate BETWEEN with WHERE (Authors born between {year_start} and {year_end})\nSQL:\n" + query_11)
print("Results (first 5):", results_11[:5])
print("\n")

# Query 12: Demonstrate COUNT with JOIN and WHERE - Count books by a specific first name and last name of an author.
author_first = 'Alice'
author_last = 'Smith'
query_12 = f"""SELECT COUNT(b.book_id) FROM Books b JOIN Authors a ON b.author_id = a.author_id WHERE a.first_name = '{author_first}' AND a.last_name = '{author_last}';"""
cursor.execute(query_12)
results_12 = cursor.fetchone()[0]
print(f"Query 12: Demonstrate COUNT with JOIN and WHERE (Books by {author_first} {author_last})\nSQL:\n" + query_12)
print("Results:", results_12)
print("\n")

# Close the connection
conn.close()
print("Database connection closed.")
```
